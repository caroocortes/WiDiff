```python
import json

with open('loading_creation_stats.json', 'r') as f:
    json_stats = json.load(f)

postgresql_stats = json_stats['postgresql']

sum_loading_time = 0
sum_total_rows = 0
for table, stats in postgresql_stats.items():
    if table != 'preflight' and table != 'index_creation_time_sec':
        sum_loading_time += stats['load_time_sec']
        sum_total_rows += tuple(stats.values())[0]

print('LOADING AND INDEX CREATION STATS')
print('[POSTGRESQL]')
print(f"Total loading time: {sum_loading_time:.2f} seconds")
print(f"Index creation time: {postgresql_stats['index_creation_time_sec']:.2f} seconds")
print(f"Total loading + index creation time: {sum_loading_time + postgresql_stats['index_creation_time_sec']:.2f} seconds")
print(f"Total rows: {sum_total_rows:,}")

print()
print('[QLEVER]')

sum_all_triples = 0
sum_ser_time = 0
triples_creation_stats = json_stats['triples_creation']
for table, stats in triples_creation_stats.items():
    sum_all_triples += stats['num_triples']
    sum_ser_time += stats['serialization_time_sec']

print(f"Total triples: {sum_all_triples:,}")
print(f"Total serialization time: {sum_ser_time:.2f} seconds")
print(f"Total loading + index creation time: {json_stats['qlever']['loading_index_creation_time_sec']:.2f} seconds")
print(f'Loading + index creation + serialization time: {json_stats["qlever"]["loading_index_creation_time_sec"] + sum_ser_time:.2f} seconds')
```
