```python
import pandas as pd
import os

datatypes_rb = ['time', 'quantity', 'globecoordinate_latitude', 'globecoordinate_longitude', 'entity', 'text']

for datatype in datatypes_rb:
    if os.path.exists('rb_samples_{}.csv'.format(datatype)):
        print(f"Processing {datatype.upper()}")
        df = pd.read_csv('rb_samples_{}.csv'.format(datatype))
    else:
        continue

    def normalize_label(val):
        if pd.isna(val):
            return val
        parts = [p.strip() for p in str(val).split(',')]
        return ','.join(sorted(parts))

    df[('label_norm')] = df['label'].apply(normalize_label)
    df['correct_label_norm'] = df['correct_label'].apply(normalize_label)

    df['match'] = df['label_norm'] == df['correct_label_norm']

    # overall
    print(f"\nOverall: {df['match'].sum()} / {len(df)} correct "
        f"({100 * df['match'].mean():.2f}%)")
    
    if 'branch' in df.columns:
        per_branch = df.groupby('branch')['match'].agg(['sum', 'count', 'mean'])
        per_branch['mean'] = (per_branch['mean'] * 100).round(2)
        print("\nPer branch:")
        print(per_branch)

    mismatches = df[~df['match']]
    print(f"\n{len(mismatches)} mismatches:")
    print(mismatches[['old_value', 'new_value', 'label', 'correct_label']].to_string())
```
